```python
# Cell 1 — Import Libraries
import os
import yaml
import glob
from pathlib import Path
from collections import defaultdict

print("✅ Libraries imported successfully")
```

```python
# Cell 2 — Create data.yaml for YOLOv8
output_path = "../output/final_dataset"

data_yaml = {
    'path': os.path.abspath(output_path),
    'train': 'train/images',
    'val': 'val/images',
    'test': 'test/images',
    'nc': 6,
    'names': ['Plastic', 'Glass', 'Metal', 'Paper', 'Organic', 'Non-Recyclable']
}

yaml_path = os.path.join(output_path, 'data.yaml')
with open(yaml_path, 'w') as f:
    yaml.dump(data_yaml, f, default_flow_style=False, allow_unicode=True)

print("✅ data.yaml created successfully")
print(f"\n📄 المحتوى:\n")
with open(yaml_path, 'r') as f:
    print(f.read())
```

```python
# Cell 3 — Final Summary
print("🎉 الداتا جاهزة للـ Training!\n")
print("="*50)

total = 0
for split in ['train', 'val', 'test']:
    imgs = glob.glob(f"{output_path}/{split}/images/*")
    total += len(imgs)
    print(f"  {split:<10} {len(imgs):>8} صورة")

print(f"  {'الإجمالي':<10} {total:>8} صورة")
print("="*50)

print(f"""
📁 Structure:
  final_dataset/
  ├── data.yaml        ← YOLOv8 config
  ├── train/           ← {len(glob.glob(f"{output_path}/train/images/*")):,} صورة
  ├── val/             ← {len(glob.glob(f"{output_path}/val/images/*")):,} صورة
  └── test/            ← {len(glob.glob(f"{output_path}/test/images/*")):,} صورة

🎯 Classes:
  0: Plastic
  1: Glass
  2: Metal
  3: Paper
  4: Organic
  5: Non-Recyclable

✅ الداتا جاهزة، الخطوة الجاية: Training على YOLOv8
""")
```
